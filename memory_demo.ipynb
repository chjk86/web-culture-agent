```python
import os
from dotenv import load_dotenv
load_dotenv(override=True)

from mem0 import Memory
os.environ["MEM0_TELEMETRY"] = "false"
os.environ["GEMINI_API_KEY"] = os.environ["GOOGLE_API_KEY"]
```

```python
import time

unique_path = f"/tmp/qdrant_{int(time.time())}"

config = {
    "llm": {
        "provider": "gemini",
        "config": {
            "model": "gemini-flash-latest",
            "temperature": 0.2,
        }
    },
    "embedder": {
        "provider": "fastembed",
        "config": {
            "model": "BAAI/bge-small-en-v1.5",
        }
    },
    "vector_store": {
        "config": {
            "embedding_model_dims": 384,
            "path": unique_path,
        }
    },
    "custom_instructions": "저장할 사실은 반드시 한국어 문장으로 기록하세요. 시점 정보가 있다면 포함하세요.",
}
memory = Memory.from_config(config)
print("사용 경로:", unique_path)
```

```python
user_id = "professor_kim"
memory.add("이 학생은 통계학 기초를 특히 어려워한다", user_id=user_id)
memory.add("2학년 전공필수 수업을 담당한다", user_id=user_id)

results = memory.search("이 학생에게 어떤 자료를 추천할까?", filters={"user_id": user_id})
for r in results["results"]:
    print("-", r["memory"])
```
