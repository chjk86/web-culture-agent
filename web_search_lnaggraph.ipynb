```python
import os
from dotenv import load_dotenv
load_dotenv(override=True)

print("TAVILY_API_KEY 앞 6자리:", os.environ.get("TAVILY_API_KEY", "없음")[:6])
```

```python
from tavily import TavilyClient

tavily_client = TavilyClient(api_key=os.environ["TAVILY_API_KEY"])

response = tavily_client.search(
    query="옥씨부인전 드라마 평론 반응",
    max_results=5,
    search_depth="basic",
    topic="general",
    include_answer=False,
    include_raw_content=False,
)

for r in response["results"]:
    print("제목:", r["title"])
    print("URL:", r["url"])
    print("관련도 점수:", r["score"])
    print("내용 조각:", r["content"][:200])
    print("---")
```

```python
response_advanced = tavily_client.search(
    query="옥씨부인전 드라마 평론 반응",
    max_results=5,
    search_depth="advanced",
    topic="general",
    include_answer=False,
    include_raw_content=True,
    include_domains=[
        "artinsight.co.kr",
        "ize.co.kr",
        "cine21.com",
        "hankyung.com",
        "khan.co.kr",
        "hani.co.kr",
        "chosun.com",
        "joongang.co.kr",
        "yna.co.kr",
        "newsis.com",
    ],
)

for r in response_advanced["results"]:
    print("제목:", r["title"])
    print("URL:", r["url"])
    print("관련도 점수:", r["score"])
    raw = r.get("raw_content", "")
    print("본문(앞 500자):", raw[:500] if raw else "(raw_content 없음)")
    print("---")
```

```python
queries = [
    "옥씨부인전 평론",
    "옥씨부인전 비평 분석",
    "옥씨부인전 신분 서사 여성 연대",
    "옥씨부인전 퀴어 재현",
]

all_results = []
for q in queries:
    res = tavily_client.search(
        query=q,
        max_results=5,
        search_depth="basic",
        topic="general",
        include_answer=False,
        include_raw_content=False,
    )
    all_results.extend(res["results"])

# 중복 URL 제거
seen_urls = set()
unique_results = []
for r in all_results:
    if r["url"] not in seen_urls:
        seen_urls.add(r["url"])
        unique_results.append(r)

print(f"총 {len(unique_results)}개 고유 결과")
```

```python
exclude_domains = ["namu.wiki", "wikipedia.org", "blog.naver.com", "youtube.com"]
review_keywords = ["평론", "비평", "리뷰", "분석", "칼럼", "시선", "재현"]

candidates = [
    r for r in unique_results
    if not any(d in r["url"] for d in exclude_domains)
    and (any(k in r["title"] for k in review_keywords) or r["score"] > 0.5)
]

for r in sorted(candidates, key=lambda x: x["score"], reverse=True):
    print(f"[{r['score']:.2f}] {r['title']}")
    print(r["url"])
    print("---")
```

```python
top_urls = [r["url"] for r in candidates[:3]]

extracted = tavily_client.extract(urls=top_urls)
for item in extracted["results"]:
    print("URL:", item["url"])
    print("본문:", item["raw_content"][:1000])
    print("=" * 50)
```
